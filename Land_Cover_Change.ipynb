```python
import matplotlib.pyplot as plt
import numpy as np

# Categories and values for land cover change (1990-2020)
categories = ['Grassland', 'Bare Areas', 'Water Bodies', 'Broadleaved Deciduous Forests', 'Shrublands']
changes_1990_2000 = [20, 15, -10, -25, 5]  # Example changes in land cover area (in percentage)
changes_2000_2010 = [-15, 20, -5, -20, 10]
changes_2010_2020 = [10, -5, 15, -10, -5]

# Set up bar width and positions
bar_width = 0.2
x = np.arange(len(categories))

# Create bar chart
fig, ax = plt.subplots(figsize=(10, 6))

bars1 = ax.bar(x - bar_width, changes_1990_2000, bar_width, label='1990-2000', color='blue')
bars2 = ax.bar(x, changes_2000_2010, bar_width, label='2000-2010', color='green')
bars3 = ax.bar(x + bar_width, changes_2010_2020, bar_width, label='2010-2020', color='red')

# Labels and title
ax.set_xlabel('Land Cover Class')
ax.set_ylabel('Percentage Change in Area')
ax.set_title('Land Cover Change Over Time (1990-2020)')
ax.set_xticks(x)
ax.set_xticklabels(categories, rotation=30)
ax.legend()

# Show the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Provided land cover percentage values from 1990 to 2020
land_cover_data = {
    "Class": ["Cropland, rainfed", "Shrubland", "Grassland", "Bare areas", "Water bodies", "Tree cover, broadleaved"],
    "1990": [8.582, 14.998, 10.926, 15.346, 6.378, 0.030],
    "2000": [8.622, 14.998, 10.715, 15.346, 6.378, 0.030],
    "2010": [8.573, 14.998, 10.835, 15.606, 6.378, 0.030],
    "2020": [8.573, 14.998, 10.795, 15.586, 6.378, 0.030]
}

# Convert to DataFrame
df = pd.DataFrame(land_cover_data)

# Compute percentage changes for each decade
df["Change_1990_2000"] = ((df["2000"] - df["1990"]) / df["1990"]) * 100
df["Change_2000_2010"] = ((df["2010"] - df["2000"]) / df["2000"]) * 100
df["Change_2010_2020"] = ((df["2020"] - df["2010"]) / df["2010"]) * 100

# Display the results
#import ace_tools as tools
#tools.display_dataframe_to_user(name="Land Cover Percentage Changes", dataframe=df)

#import ace_tools as tools
#tools.display_dataframe_to_user(name="Filtered NDVI Dataset (Outliers Removed)", dataframe=filtered_ndvi_df)
import ace_tools_open as tools; tools.display_dataframe_to_user(name="Land Cover Percentage Changes", dataframe=df)
```
